```python
# from machine_lib import *
import machine_lib as mlb

#  next step: get the access to https://platform.worldquantbrain.com/learn/operators

username,password = mlb._load_credentials()
print(f"Username: {username}, Password: {password}")   

session = mlb.login()

print(f"Session: {session}")

```

```python
url = "https://api.worldquantbrain.com/operators"

# 发送请求（确保你的 session 已经通过了 authentication 登录认证）
operators_response = session.get(url)

```

```python
if operators_response.status_code == 200:
    # 核心修改 2：直接使用 .json() 解析，将其转换为 Python 的字典或列表
    operators_data = operators_response.json()
    
    # 兼容处理：WQ Brain 的 API 格式有时直接返回列表，有时返回带 'results' 键的字典
    if isinstance(operators_data, dict):
        operators_list = operators_data.get("results", [])
    else:
        operators_list = operators_data

    print(f"成功获取到 {len(operators_list)} 个算子。前3个示例：")
    for op in operators_list[:]:
        # 这里的 op 通常是一个字典，包含 name, description 等字段
        print(op)
        
else:
    print(f"请求失败，状态码: {operators_response.status_code}")
    print(operators_response.text)
```

```python
import pandas as pd

opes = pd.DataFrame(operators_list)

print(f"Operators DataFrame:\n{opes.head()}")
```

```python
for op in operators_list:
    # if op.get("category") == "Arithmetic":
    if op["category"] == "Arithmetic":
        print(f"Arithmetic Operator: {op.get('name')} - {op.get('description')}")
```

```python
sorted_opes = opes.sort_values(by="category")
for category, group in sorted_opes.groupby("category"):
    print(f"Category: {category}")
    print("\n")
    print("-----------------------------------------------------------")
   
    for _, row in group.iterrows():
        print(f"  Operator: {row['name']} ------ {row['definition']}")
        print(f"  Description: {row['description']}")
        print("-----------------------------------------------------------")
    
    print("\n\n")
```
